```python
import matplotlib.pyplot as plot
import pandas as pd
df['Medicine Dose (mg)'] = pd.to_numeric(df['Medicine Dose (mg)'], errors='coerce')
df_cleaned = df.dropna(subset=['Recovery Days', 'Medicine Dose (mg)'])
plot.scatter(df_cleaned['Recovery Days'], df_cleaned['Medicine Dose (mg)'])
plot.show()
```

```python
import matplotlib.pyplot as plot
plot.title("Recovery Days vs Medicine Dose")
plot.xlabel("Recovery Days")
plot.ylabel("Medicine Dose(mg)")
plot.bar(df['Recovery Days'],df['Medicine Dose (mg)'],color='purple')
plot.show()
```

```python
plt.hist(df['Recovery Days'],color="Green",bins=4)
plt.show()
```

```python
import matplotlib.pyplot as plot
plot.title("Recovery Days and Medicine Dose")
plot.xlabel("Recovery Days")
plot.ylabel("Medicine Dose(mg)")
plot.plot(df['Recovery Days'],df['Medicine Dose (mg)'],color='purple')
plot.show()
```

```python
plt.title("Recovery Days and Medicine Dose")
plt.pie(df['Recovery Days'],labels=df['Medicine Dose (mg)'],autopct='%1.1f%%')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.suptitle("DASHBOARD")
plt.subplot(2, 2, 1)
plt.bar(df['Recovery Days'], df['Medicine Dose (mg)'], color='purple')
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.subplot(2, 2, 2)
plt.plot(df['Recovery Days'],df['Medicine Dose (mg)'])
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.subplot(2, 2, 3)
plt.pie(df['Recovery Days'], labels=df['Medicine Dose (mg)'], autopct='%1.1f%%')
plt.tight_layout()
plt.subplot(2, 2, 4)
plt.scatter(df['Recovery Days'], df['Medicine Dose (mg)'])
plt.subplot(2, 2, 4)
plt.hist(df['Recovery Days'], color='Green', bins=4)
plt.tight_layout()
plt.savefig('dashboard.png')
plt.show()
```
